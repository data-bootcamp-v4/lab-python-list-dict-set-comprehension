````python
products = ["t-shirt", "mug", "hat", "book", "keychain"]

def initialize_inventory(list):

    inventory = {key: int(input(f"Enter que quantity of {key} available: ")) for key in list}

    return inventory
````

````python
def get_customer_orders():
  number_orders = int(input("Enter the number of customer orders: "))
  customer_orders = {input("Enter the name of a product that a customer wants to order: ") for i in range(number_orders)}



  return customer_orders

customer_orders = get_customer_orders()
````

````python
def total_price():
  total_product_ordered = len([product for product in customer_orders])
  price_product = (sum([int(input(f"Enter the price of {product}: ")) for product in customer_orders]))

  return f"Total Price: {price_product}"
````

````python
def update_inventory(inventory):

    inv = {key: value - 1 if key in customer_orders else value for key, value in inventory.items()}


    return inv
````
