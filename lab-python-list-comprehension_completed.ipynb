````python
# List of available products
products = ["t-shirt", "mug", "hat", "book", "keychain"]


def initialize_inventory(products):
    """Ask for the available quantity of each product."""

    inventory = {
        product: int(input(f"Enter the quantity of {product}s available: "))
        for product in products
    }

    return inventory 


def get_customer_orders():
    """Ask how many products the customer wants and collect their names."""

    number_of_orders = int(
        input("Enter the number of customer orders: ")
    )

    customer_orders = {
        input(
            "Enter the name of a product that a customer wants to order: "
        ).strip().lower()
        for _ in range(number_of_orders)
    }

    return customer_orders


def calculate_order_statistics(customer_orders, products):
    """Calculate the number and percentage of unique products ordered."""

    total_products_ordered = len(customer_orders)

    percentage_unique_ordered = (
        total_products_ordered / len(products)
    ) * 100

    return total_products_ordered, percentage_unique_ordered


def update_inventory(inventory, customer_orders):
    """Subtract ordered products and remove products with zero quantity."""

    updated_inventory = {
        product: quantity - 1
        if product in customer_orders
        else quantity
        for product, quantity in inventory.items()
    }

    # Keep only products whose remaining quantity is greater than zero
    updated_inventory = {
        product: quantity
        for product, quantity in updated_inventory.items()
        if quantity > 0
    }

    return updated_inventory


def calculate_total_price(customer_orders):
    """Ask for each ordered product's price and calculate the total."""

    prices = {
        product: float(input(f"Enter the price of {product}: "))
        for product in customer_orders
    }

    total_price = sum(prices.values())

    return total_price


def print_updated_inventory(inventory):
    """Print every product remaining in the inventory."""

    inventory_lines = [
        f"{product}: {quantity}"
        for product, quantity in inventory.items()
    ]

    print("\n".join(inventory_lines))


# Run the program
inventory = initialize_inventory(products)

customer_orders = get_customer_orders()

total_products_ordered, percentage_unique_ordered = (
    calculate_order_statistics(customer_orders, products)
)

updated_inventory = update_inventory(
    inventory,
    customer_orders
)

print("\nOrder Statistics:")
print(f"Total Products Ordered: {total_products_ordered}")
print(
    f"Percentage of Unique Products Ordered: "
    f"{percentage_unique_ordered:.1f}%"
)

print("\nUpdated Inventory:")
print_updated_inventory(updated_inventory)

total_price = calculate_total_price(customer_orders)

print(f"Total Price: {total_price:.1f}")
````
