## Exercise: Managing Customer Orders Optimized with Comprehension

````python
# Step 1: Define products
products = ["t-shirt", "mug", "hat", "book", "keychain"]

# Step 2: Initialize inventory using comprehension
def initialize_inventory(products):
    return {
        product: int(input(f"Enter the quantity of {product}s available: "))
        for product in products
    }

# Step 3: Get customer orders using comprehension
def get_customer_orders():
    num_orders = int(input("Enter the number of customer orders: "))
    return {
        input("Enter the name of a product that a customer wants to order: ")
        for _ in range(num_orders)
    }

# Step 4: Calculate order statistics
def calculate_order_statistics(customer_orders, products):
    total_products_ordered = len(customer_orders)
    percentage_ordered = (total_products_ordered / len(products)) * 100
    return total_products_ordered, percentage_ordered

# Step 5: Update inventory using comprehension
def update_inventory(inventory, customer_orders):
    # subtract 1 if product is ordered
    updated_inventory = {
        product: quantity - 1 if product in customer_orders else quantity
        for product, quantity in inventory.items()
    }
    # remove items with 0 quantity
    return {
        product: quantity
        for product, quantity in updated_inventory.items()
        if quantity > 0
    }

# Step 6: Calculate total price using comprehension
def calculate_total_price(customer_orders):
    prices = {
        product: float(input(f"Enter the price of {product}: "))
        for product in customer_orders
    }
    return sum(prices.values())

# ---- MAIN PROGRAM ----

inventory = initialize_inventory(products)
customer_orders = get_customer_orders()

# Order statistics
total, percentage = calculate_order_statistics(customer_orders, products)

print("\nOrder Statistics:")
print(f"Total Products Ordered: {total}")
print(f"Percentage of Unique Products Ordered: {percentage}")

# Update inventory
inventory = update_inventory(inventory, customer_orders)

print("\nUpdated Inventory:")
for product, quantity in inventory.items():
    print(f"{product}: {quantity}")

# Total price
total_price = calculate_total_price(customer_orders)
print(f"\nTotal Price: {total_price}")
````
