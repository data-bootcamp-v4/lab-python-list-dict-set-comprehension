## 1. intialize inventory using dict.

```python
def initialize_inventory(products):
    inventory = {product: int(input(f"Enter the quantity of {product}s available: ")) for product in products}
    return inventory

# Example usage:
products = ["t-shirt", "mug", "hat", "book", "keychain"]
inventory = initialize_inventory(products)
print("\nInventory initialized:")
print(inventory)
```

## 2. get customer order

```python
def get_customer_orders(products):
    customer_orders = set()
    num_orders = int(input("Enter the number of customer orders: "))
    
    for i in range(num_orders):
        while True:
            order = input(f"Enter the name of a product that a customer wants to order: ").lower()
            if order not in products:
                print("Invalid product. Choose from the available products.")
            elif order in customer_orders:
                print("You already added this product.")
            else:
                customer_orders.add(order)
                break
    return customer_orders

# Example usage:
customer_orders = get_customer_orders(products)
print("\nCustomer Orders:")
print(customer_orders)
```

## 3. update ineventory using compreshension 

```python
def update_inventory(customer_orders, inventory):
    # Subtract 1 for each ordered product
    for item in customer_orders:
        if item in inventory and inventory[item] > 0:
            inventory[item] -= 1

    # Remove products with 0 quantity using dict comprehension
    inventory = {product: qty for product, qty in inventory.items() if qty > 0}
    return inventory

# Example usage:
inventory = update_inventory(customer_orders, inventory)
print("\nUpdated Inventory:")
print(inventory)
```

## 4, Calculate order 

```python
def calculate_order_statistics(customer_orders, products):
    total_ordered = len(customer_orders)
    percentage_ordered = (total_ordered / len(products)) * 100
    return total_ordered, percentage_ordered

# Example usage:
order_stats = calculate_order_statistics(customer_orders, products)
print("\nOrder Statistics:")
print(f"Total Products Ordered: {order_stats[0]}")
print(f"Percentage of Unique Products Ordered: {order_stats[1]:.2f}%")
```

# claculate total price of customer order 

```python
def calculate_total_price(customer_orders):
    total_price = sum([float(input(f"Enter the price of {product}: ")) for product in customer_orders])
    return total_price

# Example usage:
total_price = calculate_total_price(customer_orders)
print(f"\nTotal Price: {total_price}")
```
