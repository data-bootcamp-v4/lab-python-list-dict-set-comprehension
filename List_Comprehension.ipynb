```python
#  Initialize Inventory
def initialize_inventory(products):
    inventory = {}
    print("=== Initialize Inventory ===")
    for product in products:
        while True:
            try:
                qty = int(input(f"Enter the quantity of {product}s available: "))
                if qty < 0:
                    print("Quantity cannot be negative. Try again.")
                    continue
                inventory[product] = qty
                break
            except ValueError:
                print("Invalid input. Enter a number.")
    return inventory


#  Get Customer Orders
def get_customer_orders(products):
    print("\n=== Customer Orders ===")
    while True:
        try:
            count = int(input("Enter the number of customer orders: "))
            if count < 0:
                print("Number of orders cannot be negative.")
                continue
            break
        except ValueError:
            print("Invalid input. Enter a number.")

    customer_orders = set()
    for _ in range(count):
        product = input("Enter product name: ").strip().lower()
        if product in products:
            customer_orders.add(product)
        else:
            print("Invalid product, ignored.")

    return customer_orders


# Update Inventory to Remove products with 0 quantity
def update_inventory(customer_orders, inventory):
    for product in customer_orders:
        if product in inventory:
            inventory[product] -= 1

    inventory = {product: qty for product, qty in inventory.items() if qty > 0}
    return inventory


#  Calculate Order Statistics
def calculate_order_statistics(customer_orders, products):
    total_products_ordered = len(customer_orders)
    percentage_ordered = (total_products_ordered / len(products)) * 100
    return total_products_ordered, percentage_ordered


# Print Order Statistics
def print_order_statistics(order_statistics):
    print("\nOrder Statistics:")
    print(f"Total Products Ordered: {order_statistics[0]}")
    print(f"Percentage of Unique Products Ordered: {order_statistics[1]:.1f}")


#  Print Updated Inventory
def print_updated_inventory(inventory):
    print("\n=== Updated Inventory ===")
    if not inventory:
        print("All products are sold out.")
    else:
        for product, qty in inventory.items():
            print(f"{product}: {qty}")


# Calculate Total Price Using
def calculate_total_price(customer_orders):
    print("\n=== Calculate Total Price ===")
    prices = {}
    for product in customer_orders:
        while True:
            try:
                price = float(input(f"Enter the price of {product}: "))
                if price < 0:
                    print("Price cannot be negative.")
                    continue
                prices[product] = price
                break
            except ValueError:
                print("Invalid input. Enter a number.")
    return sum(prices.values())



products = ["t-shirt", "mug", "hat", "book", "keychain"]

# Step 1
inventory = initialize_inventory(products)

# Step 2
customer_orders = get_customer_orders(products)

# Step 3
order_statistics = calculate_order_statistics(customer_orders, products)
print_order_statistics(order_statistics)

# Step 4
inventory = update_inventory(customer_orders, inventory)
print_updated_inventory(inventory)

# Step 5
total_price = calculate_total_price(customer_orders)
print(f"Total Price: {total_price}")
```
